# Ce patterns in low-Ia stars

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import paths
from sample import abundance_ratio

plt.style.use(paths.styles / 'notebook.mplstyle')
```

```python
data = pd.read_csv(paths.data / 'sample.csv')
data = data[
    (data['lim_ce_h_flag'] == 0) &
    (data['z_max'] < 3) &
    (data['Rg'] > 3) &
    (data['Rg'] < 20)
]
data['mn_mg'], data['e_mn_mg'] = abundance_ratio(data, 'mn', 'mg')
data['al_fe'], data['e_al_fe'] = abundance_ratio(data, 'al', 'fe')
low_ia = data[data['high_alpha']]
```

```python
fig, ax = plt.subplots(figsize=(3.25, 2), constrained_layout=True)
xlim = (-0.7, 0.3)
ylim = (-0.4, 0.2)
hexbin_kwargs = dict(gridsize=30, linewidths=0.2, extent=[xlim[0], xlim[1], ylim[0], ylim[1]])
pc0 = ax.hexbin(data['mg_h'], data['fe_mg_corr'], cmap='binary', **hexbin_kwargs)
pc1 = ax.hexbin(low_ia['mg_h'], low_ia['fe_mg_corr'], cmap='Reds', mincnt=1, **hexbin_kwargs)
fig.colorbar(pc0, ax=ax, label='All stars')
fig.colorbar(pc1, ax=ax, label='Low-Ia stars')
ax.set_xlabel('[Mg/H]')
ax.set_ylabel('[Fe/Mg]')
ax.set_xlim(xlim)
ax.set_ylim(ylim)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(3.25, 2), constrained_layout=True)
xlim = (-0.7, 0.3)
ylim = (-0.9, 0.5)
hexbin_kwargs = dict(gridsize=30, linewidths=0.2, extent=[xlim[0], xlim[1], ylim[0], ylim[1]])
pc0 = ax.hexbin(data['mg_h'], data['ce_mg_corr'], cmap='binary', **hexbin_kwargs)
ax.scatter(low_ia['mg_h'], low_ia['ce_mg_corr'], c='r', s=1, marker='.', rasterized=True, edgecolors='none')
pc1 = ax.hexbin(low_ia['mg_h'], low_ia['ce_mg_corr'], cmap='Reds_r', mincnt=30, **hexbin_kwargs)
fig.colorbar(pc0, ax=ax, label='All stars')
fig.colorbar(pc1, ax=ax, label='Low-Ia stars')
ax.set_xlabel('[Mg/H]')
ax.set_ylabel('[Ce/Mg]')
ax.set_xlim(xlim)
ax.set_ylim(ylim)
plt.show()
```
